```python
clean["item"] = clean["item"].str.strip().str.title()
```

```python
clean = clean.drop_duplicates(subset=["date", "item", "city"])
```

```python
clean["price"] = pd.to_numeric(clean["price"].astype(str).str.replace(",", ""), errors="coerce")
```

```python
clean["price"] = clean["price"].replace(9999.0, np.nan)
```

```python
clean["date"] = pd.to_datetime(clean["date"])
```

```python
print("Shape:", clean.shape)
```

```python
item_mean= data.groupby("item")["price"].mean()
```

```python
item_mean
```

```python
item_median= data.groupby("item")["price"].median()
```

```python
item_median
```

```python
price_pivot = data.pivot_table (index="date", columns="item", values="price", aggfunc="mean")
```

```python
pct_change = (price_pivot.iloc[-1] / price_pivot.iloc[0] - 1) * 100
```

```python
pct_change
```

```python
table = pd.DataFrame ({"Mean":item_mean,"Median":item_median,"First to Last % change" : pct_change,})
```

```python
table
```

```python
staple_with = data[data["item"] == "Staple A"]["price"]
```

```python
print("Mean with outlier:", staple_with.mean())
print("Median with outlier:", staple_with.median())
staple_without = staple_with[staple_with < 500]
print("Mean without outlier:", staple_without.mean())
print("Median without outlier:", staple_without.median())
```
