# DATASET 1: Pakistani E-Commerce Data 

```python
import pandas as pd
```

```python
PakEComData = pd.read_csv("pakistani_ecommerce_dataset.csv")
```

## Q1: Which is the most popular product category?

```python
PakEComData["product_category"].value_counts ()
```

The most popular product category is Home decor with 30 orders.

## Q2: What is the average order value? 

```python
PakEComData["order_value"].mean()
```

The average order value is 2832.15

## Q3: Which city has the highest number of orders? 

```python
PakEComData.groupby("city")["user_id"].count().sort_values(ascending = False)
```

Karachi has the highest number of orders, with 24 orders

## Q4: What is the most used payment method? 

```python
PakEComData.groupby("payment_method")["user_id"].count().sort_values (ascending = False)
```

Cash on delivery is the most used payment method with 29 order.

## Q5: Draw a bar chart showing average order value by city. 

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
city_avg_ord_val = PakEComData.groupby("city")["order_value"].mean().sort_values (ascending = False)
city_avg_ord_val
```

```python
categories =city_avg_ord_val.index
values=city_avg_ord_val.values
colors = ["skyblue", "salmon", "lightgreen", "plum", "bisque"]
plt.bar(categories, values, color=colors)
plt.xlabel("City")
plt.ylabel("Average Order Value")
plt.title("Average Order Value by City")
plt.show()
```

# DATASET 2:  Student Lifestyle & Study Habits 

```python
import pandas as pd
```

```python
studentlife = pd.read_csv ("student_lifestyle_dataset.csv")
```

## Q1: What is the average number of study hours? 


```python
studentlife["study_hours"].mean()
```

The average number of study hours is 4.40


## Q2: Who sleeps more on average – males or females?

```python
sleep_mean = studentlife.groupby("gender")["sleep_hours"].mean()
sleep_mean
```

```python
diff_hrs = sleep_mean["Female"] - sleep_mean["Male"]
diff_hrs
```

```python
diff_min = diff_hrs * 60
diff_min
```

Female students sleep slightly more on average at about 6.80 hours compared to Male students who sleep about 6.39 hours, having a difference of roughly 0.41 hours (24mins).

## Q3: Is there a relationship between screen time and sleep hours? 

```python
studentlife["screen_time"].corr(studentlife["sleep_hours"])
```

The correlation between screen time and sleep hours is 0.0524. Because this value is very close to 0, there is very weak or no linear relationship between the two variables. Thus, screen time does not show a linear association with the number of hours a student sleeps.

## Q4: Draw a histogram of daily transport cost. 

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
plt.hist(studentlife["daily_transport_cost"], bins=20, color="skyblue", edgecolor="navy", alpha=0.7)
plt.xlabel("Daily Transport Cost")
plt.ylabel("Frequency: Number of Students")
plt.title("Daily Transport Cost")
plt.show()
```

## Q5: Compare average internet usage by gender.

```python
int_by_gender = studentlife.groupby("gender")["internet_usage"].mean()
int_by_gender
```

```python
diff_int = int_by_gender["Male"] - int_by_gender["Female"]
diff_int
```

Male students use the internet more on average at about 6.0 compared to Female students who use about 5.32, having a difference of roughly 0.68.
