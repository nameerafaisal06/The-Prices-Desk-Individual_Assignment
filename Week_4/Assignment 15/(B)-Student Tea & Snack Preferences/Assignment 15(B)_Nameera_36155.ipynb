# Student Tea & Snack Preferences

```python
import pandas as pd
```

```python
tea = pd.read_csv("student_tea_snacks.csv")
tea.head()
```

## Q1: What is the most preferred tea type?

```python
tea["tea_type"].value_counts()
```

Green tea is the most preferred tea type with 29 responses.

## Q2: What is the average number of cups per gender?

```python
cups_by_gender = tea.groupby("gender")["cups_per_day"].mean()
cups_by_gender
```

```python
diff_by_gender = cups_by_gender["Male"] - cups_by_gender["Female"]
diff_by_gender
```

Male students drink more tea on average at 3.00 cups per day compared to Female students at about 2.60 cups per day, a difference of roughly 0.40 cups

## Q3: Make a bar chart of snack choices.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
snack_counts = tea["snack_choice"].value_counts()
snack_counts
```

```python
categories = snack_counts.index
values = snack_counts.values
colors = ["skyblue", "salmon", "lightgreen"]

plt.bar(categories, values, color=colors, width = 0.7)
plt.xlabel("-Snack Choice-")
plt.ylabel("Number of Students")
plt.title("Snack Choices Among Students")
plt.grid(axis="y")
plt.show()
```

## Q4: Compare tea preference by age group.

```python
tea.groupby("age_group")["prefers_tea"].value_counts()
```

Tea preference is fairly balanced across age groups: 18–20 has 20 Yes and 19 No, 21–23 has 15 Yes and 17 No, and 24–26 has 16 Yes and 13 No
