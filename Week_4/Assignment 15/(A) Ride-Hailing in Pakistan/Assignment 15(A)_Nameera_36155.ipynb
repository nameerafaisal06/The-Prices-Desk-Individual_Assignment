# Ride-Hailing In Pakistan

```python
import pandas as pd
```

```python
ride = pd.read_csv("ride_hailing_dataset.csv")
ride.head()
```

## Q1: What is the average fare by service type?

```python
avg_fare_serv = ride.groupby("service_type")["fare"].mean().sort_values(ascending=False)
avg_fare_serv
```

## Q2: Which city has the most trips?

```python
city_trips = ride["city"].value_counts()
city_trips
```

Islamabad has the most trips with 29 trips.

## Q3: What is the average distance per trip?

```python
avg_dist = ride["distance_km"].mean()
avg_dist
```

The average distance per trip is 9.61 km (9.613 km).

## Q4: Compare trip ratings by gender.

```python
rating_by_gender = ride.groupby("gender")["trip_rating"].mean()
rating_by_gender
```

```python
diff_rating = rating_by_gender ["Male"] - rating_by_gender ["Female"]
diff_rating
```

Male riders gave higher average trip ratings at about 3.33 compared to Female riders at about 2.82, a difference of roughly 0.50 points.

## Q5: Make a bar chart of service type vs. average fare.

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
categories = avg_fare_serv.index
values = avg_fare_serv.values
colors = ["skyblue", "salmon", "lightgreen", "plum"]

plt.bar(categories, values, color=colors)
plt.xlabel("Service Type")
plt.ylabel("Average Fare")
plt.title("Average Fare by Service Type")
plt.grid(axis="y")
plt.show()
```
