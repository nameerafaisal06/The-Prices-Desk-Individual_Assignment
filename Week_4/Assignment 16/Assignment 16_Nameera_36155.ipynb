## Data Cleaning

```python
crop_df = crop_df.dropna(subset=["YIELD_KG"])
crop_df
```

```python
crop_df["AREA_ACRES"] = crop_df["AREA_ACRES"].fillna(crop_df["AREA_ACRES"].mean())
crop_df["AREA_ACRES"]
```

```python
crop_df["DISTRICT"]=crop_df["DISTRICT"].fillna("Unknown")
crop_df["DISTRICT"]
```

```python
crop_df["FERTILIZER_USED"] = crop_df["FERTILIZER_USED"].fillna("No")
crop_df["FERTILIZER_USED"]
```

```python
crop_df["YIELD_VALUE"] = crop_df["YIELD_VALUE"].fillna(crop_df["YIELD_VALUE"].mean())
crop_df["YIELD_VALUE"]
```

```python
crop_df["HARVEST_DATE"] = pd.to_datetime(crop_df["HARVEST_DATE"], format="%d/%m/%Y", errors="coerce")
crop_df["HARVEST_DATE"]
```

```python
crop_df["FARM_ID"] = crop_df["FARM_ID"].str.replace("F", "").astype(int)
crop_df["FARM_ID"]
```

## Data Analysis and Visualization

```python
total_yield = crop_df.groupby("CROP_TYPE").agg({"YIELD_KG": "sum"})
total_yield
```

```python
categories = total_yield.index
values = total_yield["YIELD_KG"]

colors = ["skyblue", "salmon", "lightgreen","mistyrose"]
plt.bar(categories, values, color=colors, width = 0.7)
plt.xlabel("Crop Type")
plt.ylabel("Yield (KG)")
plt.title ("Total Yield by Crop Type (KG)")
plt.show()
```

```python
pivot_table = crop_df.pivot_table(values='YIELD_KG', index='HARVEST_DATE', aggfunc='sum')
pivot_table
```

```python
x = pivot_table.index
y = pivot_table.values

plt.figure(figsize=(25, 10))
plt.plot(x,y, linewidth=0.8)
plt.title ("Yield Trend Over Time (KG)")
plt.xlabel ("Harvest Time")
plt.ylabel ("Yield (KG)")
plt.savefig("crop_yield_trend.png")
plt.show()
```
