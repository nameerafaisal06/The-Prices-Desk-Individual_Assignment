# AFTER AI

```python
import pandas as pd
```

```python
import numpy as np
```

```python
raw = pd.read_csv("teaching_prices_messy.csv")
```

```python
clean = raw.copy()
```

```python
print("=== RAW BASELINE DIAGNOSTICS ===")
```

```python
print("Shape:", raw.shape)
```

```python
print("Missing Values:\n", raw.isna().sum())
```

```python
print("\nRaw Summary Stats:")
```

```python
print(raw.describe(include='all'))
```

```python
clean = clean.drop_duplicates()
```

```python
clean["item"] = clean["item"].str.strip().str.title()
```

```python
clean = clean.drop_duplicates(subset=["date", "item", "city"])
```

```python
clean["price"] = pd.to_numeric(clean["price"].astype(str).str.replace(",", ""), errors="coerce")
```

```python
clean["price"] = clean["price"].replace(9999.0, np.nan)
```

```python
clean["date"] = pd.to_datetime(clean["date"])
```

```python
print("=== CLEANED DATASET DIAGNOSTICS ===")
```

```python
print("Shape:", clean.shape)
```

```python
print("Missing Values:\n", clean.isna().sum())
```

```python
print("\nCleaned Price Describe:")
```

```python
print(clean.describe())
```

```python
print("\nItem Frequency Counts:")
```

```python
print(clean["item"].value_counts())
```

```python
clean.to_csv("teaching_prices_clean.csv", index=False)
```
